```python
import yfinance as yf

ticker = yf.Ticker("INFY.NS")
info = ticker.info
info
```

```python
price = info.get("currentPrice") or info.get("regularMarketPrice", 0.0)
prev_close = info.get("previousClose") or price

change = round(price - prev_close, 2)
change_pct = round((change / prev_close) * 100, 2)

print(f"Name       : {info.get('longName')}")
print(f"Price      : ₹{price}")
print(f"Change     : {change} ({change_pct}%)")
print(f"Volume     : {info.get('volume'):,}")
print(f"Market Cap : ₹{info.get('marketCap'):,}")
print(f"P/E Ratio  : {info.get('trailingPE')}")
print(f"52W High   : ₹{info.get('fiftyTwoWeekHigh')}")
print(f"52W Low    : ₹{info.get('fiftyTwoWeekLow')}")
print(f"Sector     : {info.get('sector')}")
```

```python
df = ticker.history(period="1mo")
df.head(10)
```

```python
df = df.reset_index()

cleaned = []
for _, row in df.iterrows():
    cleaned.append({
        "date": str(row["Date"].date()),
        "open": round(float(row["Open"]), 2),
        "high": round(float(row["High"]), 2),
        "low": round(float(row["Low"]), 2),
        "close": round(float(row["Close"]), 2),
        "volume": int(row["Volume"]),
    })

cleaned[:3]  # preview first 3 entries
```

```python
NIFTY50_TICKERS = [
    "RELIANCE.NS", "TCS.NS", "HDFCBANK.NS", "INFY.NS", "ICICIBANK.NS",
    "HINDUNILVR.NS", "SBIN.NS", "BHARTIARTL.NS", "ITC.NS", "KOTAKBANK.NS",
]

query = "HDFC"

results = []
for t in NIFTY50_TICKERS:
    if query.upper() in t:
        info = yf.Ticker(t).info
        results.append({
            "ticker": t,
            "name": info.get("longName") or info.get("shortName", t),
            "exchange": info.get("exchange", "NSE"),
        })

results
```

```python
indices = [
    {"name": "Nifty 50",   "ticker": "^NSEI"},
    {"name": "Sensex",     "ticker": "^BSESN"},
    {"name": "Bank Nifty", "ticker": "^NSEBANK"},
]

for idx in indices:
    info = yf.Ticker(idx["ticker"]).info
    price = info.get("regularMarketPrice") or info.get("currentPrice", 0.0)
    prev  = info.get("regularMarketPreviousClose") or price
    pct   = round(((price - prev) / prev) * 100, 2) if prev else 0.0
    print(f"{idx['name']:12} → ₹{price:>10,.2f}  ({pct:+.2f}%)")
```

```python
SAMPLE = ["RELIANCE.NS", "TCS.NS", "INFY.NS", "HDFCBANK.NS", "SBIN.NS"]

movers = []
for t in SAMPLE:
    info = yf.Ticker(t).info
    price = info.get("currentPrice") or info.get("regularMarketPrice", 0.0)
    prev  = info.get("previousClose") or price
    pct   = round(((price - prev) / prev) * 100, 2) if prev else 0.0
    movers.append({"ticker": t, "price": price, "change_pct": pct})

movers.sort(key=lambda x: x["change_pct"], reverse=True)
print("Gainers:", movers[:2])
print("Losers: ", movers[-2:])
```
